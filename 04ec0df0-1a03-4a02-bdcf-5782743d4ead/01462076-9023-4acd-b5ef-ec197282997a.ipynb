## 5단계. 파일로 저장하기

```python
df.to_csv('벅스차트_TOP100.csv',       # 저장할 파일 이름
          index=False,                 # 맨 앞의 번호 열(0,1,2...)은 빼고 저장
          encoding='utf-8-sig')       # 엑셀에서 열어도 한글이 안 깨지는 인코딩

print('저장 완료!')                    # 완료 메시지 출력
```

### 💡 왜 `utf-8-sig` 인가?

- 그냥 저장하면 → **엑셀에서 열 때 한글이 깨집니다**
- `utf-8-sig` 는 파일 맨 앞에 "이 파일은 UTF-8입니다"라는 표식(BOM)을 붙여서 엑셀이 제대로 읽게 합니다
- 조금 뒤 공공데이터를 **읽을 때**도 인코딩 문제를 만납니다 — 읽기와 쓰기 모두 인코딩이 중요합니다

## ✅ 오늘 쓴 함수 정리

| 함수 | 하는 일 |
|---|---|
| `pd.read_html(주소)` | 웹페이지의 표 가져오기 |
| `df.head()` | 위 5줄 미리보기 |
| `df.info()` | 행 수·컬럼·빈 값 요약 |
| `df.shape` | (행, 열) 크기 |
| `df.dropna(axis=1, how='all')` | 전부 빈 컬럼 제거 |
| `df.columns` | 컬럼 이름 확인 |
| `df[df['컬럼'].str.contains('단어')]` | 조건에 맞는 행만 골라내기 |
| `df.to_csv(파일명, encoding='utf-8-sig')` | 한글 안 깨지게 CSV 저장 |
