# 📋 전체 코드 한눈에 보기

지금까지 나눠 실행한 것을 **하나로 이어 붙인 것**입니다. 흐름을 확인할 때 여기를 보세요.

```python
import pandas as pd

# ── 1. 가져오기 ───────────────────────────────
url = 'https://music.bugs.co.kr/chart'
tables = pd.read_html(url)
df = tables[0]

# ── 2. 살펴보기 ───────────────────────────────
print(df.shape)
print(df.columns)

# ── 3. 정리하기 ───────────────────────────────
df = df.dropna(axis=1, how='all')

# ── 4. 골라내기 ───────────────────────────────
결과 = df[df['아티스트'].str.contains('아이유')]
print(결과)

# ── 5. 저장하기 ───────────────────────────────
df.to_csv('벅스차트_TOP100.csv', index=False, encoding='utf-8-sig')
print('저장 완료')
```

## ✅ 오늘 사용한 함수

| 함수 | 하는 일 |
|---|---|
| `pd.read_html(주소)` | 웹페이지의 표 가져오기 |
| `len(tables)` · `tables[0]` | 표 개수 확인 · 첫 번째 선택 |
| `df.head()` · `df.info()` · `df.shape` | 표 살펴보기 |
| `df.columns` | 열 이름 확인 |
| `df.dropna(axis=1, how='all')` | 전부 빈 열 제거 |
| `df[df['열'].str.contains('단어')]` | 조건에 맞는 행 추출 |
| `df.to_csv(파일, encoding='utf-8-sig')` | 한글 안 깨지게 저장 |
